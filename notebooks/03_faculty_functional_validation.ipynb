## Sample Size & Coverage Modeling (Thesis §3.8.2)
The evaluation uses $n = 15$ faculty evaluators across $\ge 5$ academic departments.
Based on Nielsen (1994) and Faulkner (2003) binomial discovery model:
$$P(\text{problem detection}) = 1 - (1 - p)^n$$
For an individual discovery rate $p = 0.20$ and $n = 15$ evaluators, expected usability/error discovery coverage is **96.5%**.

```python
evaluators = [1, 3, 5, 10, 15, 20]
coverage_df = pd.DataFrame([
    {
        'Evaluator Count (n)': n,
        'Formula P(detection)': f'{compute_faulkner_problem_discovery_rate(n, 0.20) * 100:.2f}%',
        'Design Target (§3.8.2)': 'Target Sample' if n == 15 else 'Reference'
    }
    for n in evaluators
])
coverage_df
```

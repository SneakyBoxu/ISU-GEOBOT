## Mathematical Formulations (Thesis §3.8.1)
1. **Context Precision@K:**
$$\text{CP@K} = \frac{\sum_{k=1}^K (\text{Precision@k} \times v_k)}{\text{Total Relevant Items in Top } K}$$
2. **Context Recall:**
$$\text{Context Recall} = \frac{|\text{Retrieved Ground Truth Sentences}|}{|\text{Total Ground Truth Sentences}|}$$
3. **Faithfulness (Hallucination Detection):**
$$\text{Faithfulness} = \frac{|\text{Claims Supported by Context}|}{|\text{Total Generated Claims}|}$$
4. **Answer Relevancy:**
$$\text{Answer Relevancy} = \frac{1}{N} \sum_{i=1}^N \cos(\mathbf{E}_{g_i}, \mathbf{E}_q) = \frac{1}{N} \sum_{i=1}^N \frac{\mathbf{E}_{g_i} \cdot \mathbf{E}_q}{\|\mathbf{E}_{g_i}\| \|\mathbf{E}_q\|}$$
